```python
#introduction to slicing of list
#genral syntax of slicing: [start_value:stop_value:step_count]
students=['dinesh','javed','chandini','masooda','akila','sana','karthik','srikanth']
print(students)
```

```python
#note:last value is always exclusive to include increment index by +1
```

```python
print(students[0:2])
```
